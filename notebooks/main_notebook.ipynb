```python
import torch
from transformers import AutoTokenizer

text = "Artificial intelligence is transforming the world."
tokenizer = AutoTokenizer.from_pretrained("distilbert/distilbert-base-uncased")
inputs = tokenizer(
    text,
    return_tensors="pt",
    padding=True,
    truncation=True,
    max_length=128
)

print(inputs)
```

```python
with torch.no_grad():
    outputs = model(**inputs)
```

```python
last_hidden_state = outputs.last_hidden_state
print(last_hidden_state.shape)
```

```python
# For classification tasks, we often use the representation corresponding to the first special token, [CLS]

cls_embedding = last_hidden_state[:,0,:]
print(cls_embedding.shape)
```

```python
print("Input IDs:", inputs["input_ids"].shape)

print("Attention Mask:", inputs["attention_mask"].shape)

print("Hidden State:", last_hidden_state.shape)

print("CLS Embedding:", cls_embedding.shape)
```
